# 05 — Retention & Cohort Analysis

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path("..")
cohort = pd.read_csv(ROOT / "data/processed/cohort_retention.csv")
selected = cohort[cohort["start_term"].isin(["2024FA","2025SP","2025FA"])]

for cohort_id, g in selected.groupby("start_term"):
    plt.plot(g["relative_term"], g["retention_rate"]*100, marker="o", label=cohort_id)
plt.title("Retention by Entry Cohort")
plt.xlabel("Relative Term")
plt.ylabel("Retention Rate (%)")
plt.ylim(0,105)
plt.legend()
plt.grid(alpha=.2)
plt.show()
```
