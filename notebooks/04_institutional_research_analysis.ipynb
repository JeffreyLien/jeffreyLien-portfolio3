# 04 — Institutional Research Analysis
Creates leadership-facing KPI tables for enrollment, credit completion, withdrawals, and persistence.

```python
from pathlib import Path
import pandas as pd

ROOT = Path("..")
m = pd.read_csv(ROOT / "data/processed/student_term_metrics.csv")

annual = (
    m.groupby("academic_year")
    .agg(
        unique_students=("student_id","nunique"),
        attempted_credits=("credits_attempted","sum"),
        completed_credits=("credits_completed","sum"),
        withdrawals=("withdrawals","sum")
    )
)
annual["credit_completion_rate"] = annual["completed_credits"]/annual["attempted_credits"]
annual
```

```python
persistence = (
    m[m["eligible_for_next_term"].eq(1)]
    .groupby(["academic_year","program"])["enrolled_next_term"]
    .mean()
    .reset_index(name="next_term_persistence_rate")
)
persistence.sort_values(["academic_year","next_term_persistence_rate"]).head(12)
```
