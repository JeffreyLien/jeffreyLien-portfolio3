# 06 — Student Support & Qualitative Feedback
Combines Student Affairs operational data with open-ended student feedback.

```python
from pathlib import Path
import pandas as pd

ROOT = Path("..")
m = pd.read_csv(ROOT / "data/processed/student_term_metrics.csv")
feedback = pd.read_csv(ROOT / "data/processed/feedback_themes.csv")

eligible = m[m["eligible_for_next_term"].eq(1)].copy()
eligible["support_status"] = "No contact"
eligible.loc[(eligible["advising_contacts"]>0) & (eligible["unresolved_contacts"]==0), "support_status"] = "Contact - resolved"
eligible.loc[eligible["unresolved_contacts"]>0, "support_status"] = "Contact - unresolved"

support = eligible.groupby("support_status").agg(
    student_terms=("student_id","count"),
    observed_next_term_persistence=("enrolled_next_term","mean")
).reset_index()
support
```

```python
theme_summary = feedback.groupby("derived_theme").agg(
    responses=("response_id","count"),
    avg_satisfaction=("satisfaction_rating","mean")
).sort_values("responses", ascending=False)
theme_summary
```

Qualitative coding is transparent and auditable. A production workflow would use a formal codebook, inter-rater procedures where appropriate, and privacy controls for free-text.
