# 08 — Build Reporting Exports
Creates flat files used by the Excel reporting package and Power BI.
All Python workflow files in this repository are notebooks (`.ipynb`); there are no `.py` scripts.

```python
from pathlib import Path
import pandas as pd

ROOT = Path("..")
out = ROOT / "exports"
out.mkdir(exist_ok=True)

files = {
    "executive_student_term.csv": ROOT / "data/processed/student_term_metrics.csv",
    "program_kpis.csv": ROOT / "data/processed/program_kpis.csv",
    "course_performance.csv": ROOT / "data/processed/course_performance.csv",
    "cohort_retention.csv": ROOT / "data/processed/cohort_retention.csv",
    "student_feedback_themes.csv": ROOT / "data/processed/feedback_themes.csv",
    "external_submission_validated.csv": ROOT / "data/processed/external_submission_validated.csv",
}

for output_name, source in files.items():
    pd.read_csv(source).to_csv(out / output_name, index=False)

print(f"Created {len(files)} reporting exports in {out}")
```
