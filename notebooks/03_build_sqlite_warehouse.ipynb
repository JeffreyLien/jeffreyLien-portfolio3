# 03 — Build SQLite Institutional Research Warehouse

```python
from pathlib import Path
import pandas as pd
import sqlite3

ROOT = Path("..")
db = ROOT / "database/institutional_research.db"

with sqlite3.connect(db) as conn:
    tables = pd.read_sql_query(
        "SELECT name FROM sqlite_master WHERE type='table' ORDER BY name;", conn
    )
tables
```

Core model:
- `dim_student`
- `dim_course`
- `dim_term`
- `fact_enrollment`
- `fact_advising`
- `fact_financial_aid`
- `fact_student_feedback`
- `mart_student_term`

![Data Model](../assets/data_model.png)

```python
query = '''
SELECT
    t.academic_year,
    COUNT(DISTINCT e.student_id) AS unique_students,
    SUM(e.credits_attempted) AS attempted_credits,
    SUM(e.credits_completed) AS completed_credits
FROM fact_enrollment e
JOIN dim_term t ON e.term_id = t.term_id
GROUP BY t.academic_year
ORDER BY t.academic_year;
'''
with sqlite3.connect(db) as conn:
    annual = pd.read_sql_query(query, conn)
annual
```
